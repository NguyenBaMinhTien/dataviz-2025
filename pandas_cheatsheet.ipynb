## 1.Groupby
** Dùng khi muốn tính thống kê theo từng nhóm
** Syntax: df.groupby('cột_nhóm')['cột_tính'].hàm()

```python
import pandas as pd

# Dữ liệu điểm thi của sinh viên 3 lớp
df_diem = pd.DataFrame({
    'lop' : ['A', 'A', 'B', 'B', 'B', 'C', 'C'],
    'ten' : ['An', 'Bình', 'Chi', 'Dung', 'Em', 'Phúc', 'Giang'],
    'diem_toan' : [8.5, 7.0, 9.0, 6.5, 8.0, 7.5, 9.5],
    'diem_ly' : [7.0, 8.0, 6.0, 9.0, 7.5, 8.5, 6.5,]
})

#Tính điểm toán trung bình theo từng lớp
tb_toan = df_diem.groupby('lop')['diem_toan'].mean()
print(tb_toan)

#Tính nhiều thống kê cùng lúc
thong_ke = df_diem.groupby('lop')['diem_toan'].agg(['mean', 'max', 'min', 'count'])
print(thong_ke)
```

## 2.Merge
** Dùng khi muốn ghép 2 bảng lại theo cột chung
** Syntax: pd.merge(df1, df2, on='cột_chung', how='inner/left/right/outer')

```python
# Bảng thông tin sinh viên
df_sv = pd.DataFrame({
    'ma_sv' : [101, 102, 103, 104],
    'ten' : ['An', 'Bình', 'Chi', 'Dung']
})

# Bảng điểm
df_diem2 = pd.DataFrame({
    'ma_sv' : [101, 102, 104, 105],
    'diem' : [8.5, 7.0, 9.0, 6.5]
})

# Inner join: chỉ giữ sinh viên có trong cả 2 bảng
inner = pd.merge(df_sv, df_diem2, on='ma_sv', how='inner')
print('Inner: ')
print(inner)

#Left join: giữ tất cả sinh viên bên trái, bên phải không có thì NaN
left = pd.merge(df_sv, df_diem2, on='ma_sv', how='left')
print("Left: ")
print(left)
```

## 3.Pivot_table
** Dùng khi tạo bảng tóm tắt dạng ma trận
**Syntax : df.pivot_table(values='cột_giá_trị'), index='hàng', columns='cột', aggfunc='mean'

```python
# Dữ liệu doanh thu theo tháng và sản phẩm
df_doanhthu = pd.DataFrame({
    'thang' : ['T1','T1','T2','T2','T3','T3'],
    'san_pham' : ['A','B','A','B','A','B'],
    'doanh_thu' : [100, 150, 120, 130, 90, 200]
})

#Pivot: hàng = tháng, cột = sản phẩm, giá trị = tổng doang thu
bang = df_doanhthu.pivot_table(
    values='doanh_thu',
    index='thang',
    columns='san_pham',
    aggfunc='sum'
)
print(bang)
```

## 4.Apply
** Dùng khi muốn áp dụng một hàm tùy chỉnh lên từng hàng hoặc từng cột
** Syntax: df['cột'].apply(hàm) || df.apply(hàm, axis=1)

```python
df_hs = pd.DataFrame({
    'ten' : ['An', 'Bình', 'Chi', 'Dung'],
    'diem' : [8.5, 5.0, 6.5, 9.2]
})

## Dùng apply để xếp loại - logic tùy chỉnh, không có hàm sẵn
def xep_loai(diem):
    if diem >= 8.5:
        return 'Giỏi'
    elif diem >= 6.5:
        return 'Khá'
    else:
        return 'Trung Bình'
    
df_hs['xep_loai'] = df_hs['diem'].apply(xep_loai)
print(df_hs)
```

# 5.Melt
** Dùng khi chuyển bảng từ dạng "rộng" (wide) -> dạng "dài" (long/tidy)
Tidy data: mỗi hàng = 1 quan sát, mỗi cột = 1 biến -> dễ vẽ biểu đồ hơn
**Syntax: df.melt(id_vars=['cột_giữ'], var_name='tên_cột_mới', value_name='tên_giá_trị')


```python
# Bảng dạng rộng (wide): mỗi môn học là 1 cột
df_wide = pd.DataFrame({
    'ten' : ['An', 'Bình', 'Chi'],
    'Toan' : [8.5, 7.0, 9.0],
    'Ly' : [7.5, 8.5, 6.0],
    'Hoa' : [9.0, 6.5, 8.0]
})
print("Bảng wide: ")
print(df_wide)

# Chuyển sang dạng dài (long/tidy)
df_long = df_wide.melt(
    id_vars=['ten'],         # Cột giữ nguyên
    var_name='mon_hoc',      # Tên cột mới chứa tên môn
    value_name='diem'        # Tên cột mới chứa điểm
)
print("\nBảng long (tidy): ")
print(df_long)
```

# 6.Query
** Dùng khi lọc dữ liệu theo điều kiện
**Syntax: df.query("điều kiện")

```python
import pandas as pd

df_diem = pd.DataFrame({
    'lop' : ['A', 'A', 'B', 'B', 'B', 'C', 'C'],
    'ten' : ['An', 'Bình', 'Chi', 'Dung', 'Em', 'Phúc', 'Giang'],
    'diem_toan' : [8.5, 7.0, 9.0, 6.5, 8.0, 7.5, 9.5],
    'diem_ly' : [7.0, 8.0, 6.0, 9.0, 7.5, 8.5, 6.5]
})

# Lọc sinh viên lớp A có điểm toán > 8
result1 = df_diem.query("diem_toan > 8 and lop == 'A'")
print("Lớp A, toán > 8: ")
print(result1)

# Dùng biên bên ngoài trong query (dùng @ trước tên biến)
nguong = 8.0
result2 = df_diem.query("diem_toan > @nguong")
print(f"\nSinh viên toán > {nguong}: ")
print(result2)
```

## 7.Value_counts
** Dùng khi đếm số lần xuất hiện của từng giá trị trong 1 cột
**Syntax: df['cột].value_counts()
**Tham số hay dùng: 
- normalize=True -> ra tỉ lệ phần trăm thay vì số đếm
- dropna=False -> đếm cả ô NaN
- ascending=True -> sắp xếp tăng dần

```python
import numpy as np

df_hs = pd.DataFrame({
    'ten':    ['An', 'Bình', 'Chi', 'Dung', 'Em', 'Phúc', 'Giang', 'Hoa'],
    'lop':    ['A',  'A',   'B',   'B',    'B',  'C',    'C',     'A'],
    'xep_loai': ['Giỏi', 'Khá', 'Giỏi', 'TB', 'Khá', 'Giỏi', 'Giỏi', None]
})

# Đếm số học sinh mỗi lớp
print("Số học sinh mỗi lớp: ")
print(df_hs['lop'].value_counts())

# Ra tỉ lệ phần trăm
print("\nTỉ lệ phần trăm mỗi lớp: ")
print(df_hs['lop'].value_counts(normalize=True).round(3))

# Đếm xếp loại
print("\nXếp loại: ")
print(df_hs['xep_loai'].value_counts(dropna=False))
```

## 8.Fillna
** Dùng khi xử lý missing values (NaN), điền giá trị thay thế
**Syntax: df['cột'].fillna(giá_trị) || df.fillna(giá_trị)
**Các phương thức điền:
- Điền bằng số cố định: fillna(0)
- Điền bằng trung bình: fillna(df['cột'].mean())
- Điền bằng trung vị: fillna(df.['cột'].median())
- Điền bằng giá trị trước đó: fillna(method='ffill')

```python
df_sv = pd.DataFrame({
    'ten':  ['An', 'Bình', 'Chi', 'Dung', 'Em'],
    'tuoi': [20,   np.nan, 22,    np.nan,  21],
    'diem': [8.5,  7.0,   np.nan, 6.5,    np.nan],
    'lop':  ['A',  None,  'B',   'B',     'A']
})

print("Dữ liệu gốc: ")
print(df_sv)
print(f"\n Số NaN mỗi cột:\n{df_sv.isnull().sum()}")

# 1. Điền tuổi bằng trung bình
trung_binh_tuoi = df_sv['tuoi'].mean()
df_sv['tuoi_filled'] = df_sv['tuoi'].fillna(trung_binh_tuoi)
print(f"\nTuổi trung bình = {trung_binh_tuoi:.1f}")
print(df_sv[['ten', 'tuoi', 'tuoi_filled']])

# 2. Điền điểm bằng trung vị
trung_vi_diem = df_sv['diem'].median()
df_sv['diem_filled'] = df_sv['diem'].fillna(trung_vi_diem)
print(f"\nĐiểm trung vị = {trung_vi_diem}")
print(df_sv[['ten', 'diem', 'diem_filled']])
```
