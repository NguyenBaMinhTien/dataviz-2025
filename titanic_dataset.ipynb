```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sb

df = sb.load_dataset('titanic')
df.head()
```

** Bảng có các cột survived (0/1), pclass (1/2/3), sex, age, fare...
Cột 'survived' = 1 nghĩa là sống sót, = 0 là không sống sót.
Ô NaN ở cột 'age' → có missing values.

```python
df.tail()
df.info()
```

** 891 hàng, 15 cột. 
Các cột kiểu object: sex, embarked, class, who, embark_town, alive, alone.
Các cột kiểu float64: age, fare → có thể có NaN vì float mới lưu NaN được.

```python
df.describe()
```

Tuổi (age): từ 0.42 đến 80 tuổi, trung bình 29.7 tuổi
Giá vé (fare): từ 0 đến 512, chênh lệch rất lớn → có thể có outlier
Survived: mean = 0.38 → tỉ lệ sống sót khoảng 38%

```python
# Kiểm tra missing values

missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    'Số ô thiếu': missing_count,
    'Tỉ lệ thiếu (%)': missing_percentage
}).sort_values('Tỉ lệ thiếu (%)', ascending=False)
print(missing_df[missing_df['Số ô thiếu'] > 0])
```

- 'deck': thiếu 688/891 = 77.2%
- 'age': thiếu 177/891 = 19.9%

```python
# Phân tích cột categorical
print("sex")
print(df['sex'].value_counts())
print("\npclass")
print(df['pclass'].value_counts())
print("\nembarked")
print(df['embarked'].value_counts())
```

- sex: 577 nam (64.7%) vs 314 nữ (35.3%)
- pclass: hạng 3 đông nhất (491 người), hạng 1 ít nhất (216 người)
- embarked: S (Southampton) đông nhất với 644 người

```python
# Phân tích cột numeric
for columns in ['age', 'fare', 'sibsp']:
    print(f"\n{columns}")
    print(f"  Mean:   {df[columns].mean():.2f}")
    print(f"  Median: {df[columns].median():.2f}")
    print(f"  Min:    {df[columns].min():.2f}")
    print(f"  Max:    {df[columns].max():.2f}")
```

- fare: max = 512.33, nhưng median chỉ 14.45
- sibsp (anh chị em/vợ chồng): max = 8
- age: mean (29.7) > median (28)

```python
# 1. Phụ nữ có tỉ lệ sống sót cao hơn đàn ông không?
ti_le_gioi_tinh = df.groupby('sex')['survived'].mean()
print("Tỉ lệ sống sót theo giới tính: ")
print(ti_le_gioi_tinh)
```

- Phụ nữ: ~74% sống sót
- Đàn ông: ~19% sống sót
→ Dự đoán ĐÚNG. Chênh lệch rất lớn — gần 4 lần.
Nguyên nhân: nguyên tắc "women and children first" được áp dụng 
khi phân phối xuồng cứu sinh.

```python
# 2. Hạng vé 1 có tỉ lệ sống sót cao hơn hạng 3 không?
ti_le_hang_ve = df.groupby('pclass')['survived'].mean()
print("Tỉ lệ sống sót theo hạng vé: ")
print(ti_le_hang_ve)
```

- Hạng 1: ~63% sống sót
- Hạng 3: ~24% sống sót
→ Dự đoán ĐÚNG. Hạng vé ảnh hưởng rõ rệt.
Không chỉ vì cabin an toàn hơn mà còn vì:
khoang hạng 1 gần boong tàu hơn → tiếp cận xuồng cứu sinh dễ hơn.
Hạng 3 ở tầng dưới → khó thoát ra ngoài khi tàu chìm.

```python
# 3. Tỉ lệ sống sót toàn tàu là bao nhiêu %?
tong_nguoi = len(df)
so_nguoi_song_sot = df['survived'].sum()
so_nguoi_chet = tong_nguoi - so_nguoi_song_sot
ti_le = df['survived'].mean()

print("Tỉ lệ sống sót toàn tàu")
print(f"Tổng số hành khách : {tong_nguoi}")
print(f"Số người sống sót  : {so_nguoi_song_sot}")
print(f"Số người tử nạn    : {so_nguoi_chet}")
print(f"Tỉ lệ sống sót     : {ti_le: }")
```

- Tỉ lệ sống sót thực tế: ~38%
→ Dự đoán SAI. Tôi đoán ~20% nhưng thực tế cao hơn gần gấp đôi.
Lý do có thể sai: tưởng thảm họa khiến hầu hết không thoát được,
nhưng thực ra xuồng cứu sinh đã cứu được một phần đáng kể —
đặc biệt là phụ nữ và trẻ em.
